## LLM Answer Check
retrieval 결과를 컨텍스트로 넣어 실제 답변 품질을 확인합니다.

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from textwrap import shorten
```

```python
load_dotenv()

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
```

```python
llm_test_queries = [
    "편의점 할인 카드 추천해줘",
    "스타벅스 할인 카드 알려줘",
    "공항라운지 무료 입장 카드 알려줘",
    "해외 적립 혜택 카드 알려줘",
    "전월실적 30만원 이하 카드 알려줘",
]

llm_test_queries
```

```python
def build_context(results, max_docs=3, max_chars=500):
    parts = []
    for idx, item in enumerate(results[:max_docs], 1):
        if len(item) == 2:
            doc, score = item
        else:
            doc, score = item[0], item[1]
        parts.append(
            f"[{idx}] card_name={doc.metadata.get('card_name')} | source={doc.metadata.get('source')} | type={doc.metadata.get('type')} | quality_status={doc.metadata.get('quality_status')} | score={score:.4f}\n"
            f"{shorten(doc.page_content.replace(chr(10), ' '), width=max_chars, placeholder='...')}"
        )
    return "\n\n".join(parts)


def answer_with_retrieval(query, results, use_caution=True):
    context = build_context(results)
    caution = "review_needed 문서는 신뢰도를 낮게 보고, 명확한 근거가 없으면 추정하지 마라." if use_caution else ""
    prompt = f"""
너는 카드 혜택 검색 결과를 요약하는 도우미다.
{caution}
아래 검색 결과만 근거로 답변해라.
질문: {query}

검색 결과:
{context}

요구사항:
- 검색 결과에 근거가 있는 내용만 답변
- 카드명과 핵심 혜택을 짧게 정리
- 조건이 애매하거나 부족하면 '검색 결과만으로는 확실하지 않음'이라고 표시
"""
    return llm.invoke(prompt)
```

```python
llm_answers = {}

for query in llm_test_queries:
    results = retriever.search_with_score(query, k=3)
    response = answer_with_retrieval(query, results)
    llm_answers[query] = {
        "results": results,
        "answer": response.content,
    }
    print("=" * 100)
    print(f"[LLM QUERY] {query}")
    print("\n[TOP RESULTS]")
    print(build_context(results, max_docs=3, max_chars=250))
    print("\n[ANSWER]")
    print(response.content)
```

```python
target_llm_query = "편의점 할인 카드 추천해줘"
llm_answers.get(target_llm_query, {})
```

## Retrieval Rerank Check

벡터 검색 결과를 그대로 쓰지 않고, 질문 유형에 따라 긍정/부정 힌트를 반영해 재정렬합니다.

```python
import re

QUERY_PATTERNS = {
    "lte_30": [r"30만원 이하", r"30만 원 이하", r"전월실적 30만원 이하", r"전월 이용금액 30만원 이하"],
    "overseas_benefit": [r"해외 적립", r"해외 혜택", r"해외 할인", r"해외 마일리지"],
    "lounge": [r"공항라운지", r"공항 라운지", r"라운지 무료", r"무료 입장"],
}

POSITIVE_HINTS = {
    "lte_30": ["30만원 미만", "30만원 이하", "전월 이용금액 30만원 미만", "전월실적 30만원 미만", "실적 조건 없음", "전월실적 없음"],
    "overseas_benefit": ["적립", "마일리지", "포인트", "캐시백", "혜택"],
    "lounge": ["공항 라운지", "라운지 무료", "더라운지", "THE LOUNGE", "무료 입장"],
}

NEGATIVE_HINTS = {
    "lte_30": ["50만원 이상", "60만원 이상", "100만원 이상"],
    "overseas_benefit": ["해외이용안내", "해외 이용 안내", "수수료", "원화결제", "DCC", "청구금액 산출방법"],
    "lounge": ["유의사항", "해외이용안내"],
}

def classify_query(query):
    for label, patterns in QUERY_PATTERNS.items():
        for pattern in patterns:
            if re.search(pattern, query):
                return label
    return None

def rerank_results(query, results):
    query_type = classify_query(query)
    rescored = []

    for doc, raw_score in results:
        text = doc.page_content or ""
        adjusted_score = raw_score

        if query_type:
            for hint in POSITIVE_HINTS.get(query_type, []):
                if hint in text:
                    adjusted_score -= 0.05

            for hint in NEGATIVE_HINTS.get(query_type, []):
                if hint in text:
                    adjusted_score += 0.08

        if doc.metadata.get("quality_status") == "review_needed":
            adjusted_score += 0.03

        rescored.append((doc, raw_score, adjusted_score))

    rescored.sort(key=lambda item: item[2])
    return rescored
```

```python
reranked_results = {}

for query in test_queries:
    raw_results = retriever.search_with_score(query, k=5)
    reranked = rerank_results(query, raw_results)
    reranked_results[query] = reranked

    print("=" * 100)
    print(f"[RERANK QUERY] {query}")
    for idx, (doc, raw_score, adjusted_score) in enumerate(reranked[:3], 1):
        print(f"\n[{idx}] raw: {raw_score:.4f} / adjusted: {adjusted_score:.4f}")
        print("card_name:", doc.metadata.get("card_name"))
        print("source:", doc.metadata.get("source"))
        print("type:", doc.metadata.get("type"))
        print("quality_status:", doc.metadata.get("quality_status"))
        print("preview:", doc.page_content[:300])
        print("-" * 80)
```

## Retrieval Rerank Check (Generalized)

특정 숫자를 하드코딩하지 않고, 질문에서 `숫자 + 비교 표현`, `혜택 유형`, `안내성 문구`를 일반화해서 재정렬합니다.

```python
COMPARATORS = ("이하", "이상", "미만", "초과")

BENEFIT_TERMS = {
    "earn": ["적립", "포인트", "마일리지", "캐시백"],
    "discount": ["할인", "청구 할인", "결제일 할인"],
    "lounge": ["공항라운지", "공항 라운지", "라운지", "더라운지", "THE LOUNGE", "무료 입장"],
}

GUIDE_TERMS = ["수수료", "해외이용안내", "해외 이용 안내", "유의사항", "원화결제", "DCC", "청구금액 산출방법"]

def extract_amount_conditions(text):
    text = text.replace(" ", "")
    matches = re.findall(r"(\d+)만원(이하|이상|미만|초과)", text)
    return [(int(amount), comp) for amount, comp in matches]

def extract_benefit_intents(text):
    intents = set()
    for label, keywords in BENEFIT_TERMS.items():
        if any(keyword in text for keyword in keywords):
            intents.add(label)
    return intents

def compare_condition(query_cond, doc_cond):
    q_amount, q_comp = query_cond
    d_amount, d_comp = doc_cond

    if q_amount == d_amount and q_comp == d_comp:
        return "exact"

    if q_comp in ("이하", "미만") and d_comp in ("이상", "초과") and d_amount >= q_amount:
        return "conflict"

    if q_comp in ("이상", "초과") and d_comp in ("이하", "미만") and d_amount <= q_amount:
        return "conflict"

    if q_comp in ("이하", "미만") and d_comp in ("이하", "미만") and d_amount <= q_amount:
        return "compatible"

    if q_comp in ("이상", "초과") and d_comp in ("이상", "초과") and d_amount >= q_amount:
        return "compatible"

    return None

def generalized_rerank(query, results):
    query_conds = extract_amount_conditions(query)
    query_intents = extract_benefit_intents(query)
    rescored = []

    for doc, raw_score in results:
        text = doc.page_content or ""
        adjusted = raw_score

        doc_conds = extract_amount_conditions(text)
        for q_cond in query_conds:
            for d_cond in doc_conds:
                relation = compare_condition(q_cond, d_cond)
                if relation == "exact":
                    adjusted -= 0.08
                elif relation == "compatible":
                    adjusted -= 0.03
                elif relation == "conflict":
                    adjusted += 0.10

        if query_intents:
            doc_intents = extract_benefit_intents(text)
            for intent in query_intents:
                if intent in doc_intents:
                    adjusted -= 0.05

        if query_intents.intersection({"earn", "discount"}) and any(term in text for term in GUIDE_TERMS):
            adjusted += 0.06

        if doc.metadata.get("quality_status") == "review_needed":
            adjusted += 0.03

        rescored.append((doc, raw_score, adjusted))

    rescored.sort(key=lambda item: item[2])
    return rescored
```

```python
generalized_queries = [
    "전월실적 30만원 이하 카드",
    "전월실적 40만원 이상 카드",
    "해외 적립 혜택 카드",
    "해외 할인 혜택 카드",
    "공항라운지 무료 입장 카드",
]

generalized_queries
```

```python
import re

generalized_reranked_results = {}

for query in generalized_queries:
    raw_results = retriever.search_with_score(query, k=5)
    reranked = generalized_rerank(query, raw_results)
    generalized_reranked_results[query] = reranked

    print("=" * 100)
    print(f"[GENERALIZED RERANK QUERY] {query}")
    for idx, (doc, raw_score, adjusted_score) in enumerate(reranked[:3], 1):
        print(f"\n[{idx}] raw: {raw_score:.4f} / adjusted: {adjusted_score:.4f}")
        print("card_name:", doc.metadata.get("card_name"))
        print("source:", doc.metadata.get("source"))
        print("type:", doc.metadata.get("type"))
        print("quality_status:", doc.metadata.get("quality_status"))
        print("preview:", doc.page_content[:300])
        print("-" * 80)
```

## Current Retrieval Rule Summary

- 기본 검색은 `search_with_score()`를 사용한다.
- `search_with_selected_rerank()`는 현재 `해외 할인`, `공항라운지` 질의에만 실제 적용 대상으로 본다.
- `해외 적립`은 실험 결과가 아직 불안정해서 일반 search를 유지한다.
- `편의점`, `스타벅스` 같은 일반 혜택 질의는 기본 retrieval 결과를 그대로 사용한다.
- LLM 결과만 보지 말고, 항상 `TOP RESULTS`와 `document preview`를 같이 확인한다.

# 58 근거 보완과 답변 지침의 영향 분리

57번의 근거 판정이 완료된 뒤, 고정 12질문 × 두 운영 청킹 프로필의 순차 검색 × 근거 A/B × 기존/보완 답변 지침 × 2회 = 192개 답변을 비교한다. 별도로 원문 발췌를 제공하는 진단군 48개를 실행하며 후보 성능 분모에는 넣지 않는다. 합계 논리 요청 240개, 운영 검증 재시도 포함 최대 물리 요청 480개다.

모델 gpt-6-luna/medium, 출력 2400, 기존 스키마와 64000바이트 한도를 유지한다. 보완 지침에는 특정 질문 정답이나 카드별 수치를 넣지 않는다. 동일 질문·프로필·지침·회차에서 A/B의 실제 입력과 근거가 모두 같을 때만 A의 답변을 B에 공유하며 별도 표시한다. 서로 다른 지침이나 회차 사이에는 공유하지 않는다.

공통 지침 두 개 × benefit A/B × parent-child A/B의 전체 8설정을 공개한다. 각 프로필의 통과율 퇴행과 새로운 중요 오류·조건 손실·범위 위반을 막고 질문별 순개선이 가장 큰 설정을 고른다. 동률은 변경 축, 추가 검색 수, 입력 바이트, 실측 비용, 설정 ID 순서다. 개선이 없으면 기준 설정을 유지한다.

`.py`는 장시간 API 실행의 중단 재개, 실제 입력·지침·재시도 계약, 예산 상한 및 원시 응답 보존을 담당한다. HTML은 결과와 원문을 검토한 뒤 직접 작성한다. 답변 완료 시간은 검증 재시도를 포함하며 첫 토큰과 화면 출력 시간은 미측정이다. 운영 코드와 원본 데이터는 변경하지 않는다.

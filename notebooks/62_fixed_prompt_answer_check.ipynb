### 오프라인 판정과 코드 분리 이유

원문·실제 입력·생성 답변을 수동 비교했고 독립 검토에서 본문전용·비교카드 감사 누락을 보완했다. 카드별 본문 및 비교문을 검증하고 중립안내가 실제 혜택 주장을 숨기지 않도록 검사한다. 유료 코드셀의 동결값을 보존하면서 이 전체 검사를 반복 실행할 수 있도록 다음 작은 검증 스크립트를 분리했다. 모델 호출이나 HTML 생성 책임은 없다.

- `scripts/62_validate_answer_prose.py`: 전체 본문/카드별 조건·원문·실제 입력 결속 및 비교카드 감사.
- `scripts/62_finalize_evaluation.py`: 원본/재사용720행을 다시 확인한 후 전체 집계와 사전 선택기준 적용. `--check`는 저장결과·코드·입력 해시를 쓰기 없이 재검증한다. Python 최적화로 assertion을 제거한 실행은 거절한다.
- 기존 `scripts/62_evaluate_answers.py`와 유료 실행 코드는 동결을 보존했다. 새 전체본문 검증은 최종 집계에서 필수다.

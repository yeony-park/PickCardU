### 평가 자료 계약

`source_registry.json`의 records는 Top3 union의 질문–카드별 records다. 각 record는 query_id/card_key/source_pdf_sha256/relevant/reason/facts/core_paths를 갖는다. facts에는 condition_id/benefit_id/roles(core, extended, assertion_guard)/source_statement/source_status/source_support를 고정한다. core_paths는 같은 카드에서 모두 필요한 condition_id 목록의 허용 대안들이다.

`source_review/retrieval_judgments.json`은 모든576 identity에 실제Top3 card_reviews와 registry의 동일 source fact 전부, 카드별현재 input_status/input_quotes/input_reason을 연결한다. `source_registry_acceptance.json`은 registry_required_files로 재구성한 PDF/이미지/입력판정/jobs/검색 closure를 root가 확인한 기록이다. preanswer_closure는 집합·소유권·proof를검사하고 answer_phase_manifest를불변으로고정한다. relevant=False도평가에포함하며조회결과를걸러내지않는다.

답변판정은63형식+기존62전체본문검증형식의legacy_review다. 모든사전허용corepath를유지하고extended/guard조건은등록한role/source를참조한다. 최종평가가답변전phase를재검증한다. 이자료는평가전용이며LLM질문·근거본문에주입하지않는다. 별도보고서생성코드는없다.

```python
from pathlib import Path
import importlib.util,sys,json
ROOT=Path.cwd() if (Path.cwd()/'notebooks').is_dir() else Path.cwd().parent
spec=importlib.util.spec_from_file_location('fixed_profile_runtime',ROOT/'notebooks/scripts/63_65_experiment_runtime.py')
R=importlib.util.module_from_spec(spec);sys.modules[spec.name]=R;spec.loader.exec_module(R)
# BEGIN65_APPROVED_COST_BINDING
# Only bridge the already approved upper ledger during the existing reserve.
# Raw journals, new unknown-cost rejection, payload/cost/request caps stay intact.
def _install_approved_cost_binding65(runtime):
    if getattr(runtime.Cache, 'approved_cost_binding65', False):
        return
    prior_reserve = runtime.Cache.reserve
    legacy = runtime.X.R.P
    def reserve(self, normalized):
        if self.stage != 65:
            return prior_reserve(self, normalized)
        actual_ledger = legacy.ledger
        try:
            legacy.ledger = runtime.X.budget_upper_ledger
            return prior_reserve(self, normalized)
        finally:
            legacy.ledger = actual_ledger
    runtime.Cache.reserve = reserve
    runtime.Cache.approved_cost_binding65 = True
_install_approved_cost_binding65(R)
# END65_APPROVED_COST_BINDING

scope65=R.load('scope65_followup','65_scope_repair.py')
scope65.install(R)
queries,manifest=R.prepare(65,ROOT/'notebooks/65_fixed_profile_confirmation.ipynb')
print('단계',manifest['stage'],'질문',len(queries),'누적보수차감',R.X.budget_upper_ledger()['known_cost_usd'])
```
